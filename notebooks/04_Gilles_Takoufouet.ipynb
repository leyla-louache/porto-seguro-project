```python
from pathlib import Path
import pandas as pd
from scipy.io import arff

# ============================================================
# 1. Chemin vers ton fichier ARFF
# ============================================================

input_file = Path("dataset.arff")

# Le fichier CSV sera créé dans le même dossier
output_file = input_file.with_name("porto_seguro.csv")

# Vérifier que le fichier existe
if not input_file.exists():
    raise FileNotFoundError(
        f"Le fichier n'a pas été trouvé : {input_file.resolve()}"
    )

# ============================================================
# 2. Lire le fichier ARFF
# ============================================================

print("Lecture du fichier ARFF...")

data, metadata = arff.loadarff(input_file)

df = pd.DataFrame(data)

print("Fichier ARFF lu avec succès.")

# ============================================================
# 3. Convertir les variables catégorielles
# ============================================================

for column in df.select_dtypes(include=["object"]).columns:
    df[column] = df[column].apply(
        lambda x: x.decode("utf-8")
        if isinstance(x, bytes)
        else x
    )

# ============================================================
# 4. Informations sur le dataset
# ============================================================

print("\n========================================")
print("INFORMATIONS SUR LE DATASET")
print("========================================")

print(f"Nombre de lignes   : {df.shape[0]:,}")
print(f"Nombre de colonnes : {df.shape[1]}")

print("\nPremières lignes :")
print(df.head())

print("\n========================================")
print("CONVERSION EN CSV")
print("========================================")

# ============================================================
# 5. Sauvegarder le CSV
# ============================================================

df.to_csv(output_file, index=False)

print(f"CSV créé avec succès :")
print(output_file.resolve())

print("\n========================================")
print("TERMINÉ")
print("========================================")
```

```python
# ============================================================
# EDA ORIENTÉE TARGET - PORTO SEGURO
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------------------------------------------------
# 1. PARAMÈTRES
# ------------------------------------------------------------

# Fichier CSV créé précédemment
input_file = r"C:\Users\User\Documents\SoSe26\Advanced Data Analytics\ADA Projekt\porto_seguro.csv"

# Dossier dans lequel les graphiques seront sauvegardés
output_dir = Path(
    r"C:\Users\User\Documents\SoSe26\Advanced Data Analytics\ADA Projekt\eda_target"
)

output_dir.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. LECTURE DU DATASET
# ------------------------------------------------------------

print("Lecture du dataset...")

df = pd.read_csv(input_file)

print("Dataset chargé avec succès.")
print(f"Nombre de lignes    : {df.shape[0]:,}")
print(f"Nombre de colonnes  : {df.shape[1]}")

target = "target"

# ------------------------------------------------------------
# 3. DISTRIBUTION DE TARGET
# ------------------------------------------------------------

target_counts = df[target].value_counts().sort_index()
target_percent = df[target].value_counts(normalize=True).sort_index() * 100

print("\n" + "=" * 60)
print("DISTRIBUTION DE TARGET")
print("=" * 60)

for value in target_counts.index:
    print(
        f"target = {value}: "
        f"{target_counts[value]:,} observations "
        f"({target_percent[value]:.2f} %)"
    )

plt.figure(figsize=(9, 6))

bars = plt.bar(
    target_counts.index.astype(str),
    target_counts.values
)

plt.title("Distribution of Target Variable")
plt.xlabel("Target")
plt.ylabel("Number of observations")
plt.grid(axis="y", alpha=0.3)

for bar, count, percentage in zip(
    bars,
    target_counts.values,
    target_percent.values
):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{count:,}\n({percentage:.2f}%)",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

plt.savefig(
    output_dir / "01_target_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# 4. VARIABLES BINAIRES : TAUX DE TARGET = 1
# ------------------------------------------------------------

binary_cols = [
    col for col in df.columns
    if col.endswith("_bin")
]

print("\n" + "=" * 60)
print("ANALYSE DES VARIABLES BINAIRES")
print("=" * 60)

binary_results = []

for col in binary_cols:

    grouped = df.groupby(col)[target].mean()

    if 0 in grouped.index and 1 in grouped.index:

        rate_0 = grouped.loc[0] * 100
        rate_1 = grouped.loc[1] * 100

        difference = rate_1 - rate_0

        binary_results.append({
            "variable": col,
            "target_rate_0": rate_0,
            "target_rate_1": rate_1,
            "difference": difference
        })

binary_results = pd.DataFrame(binary_results)

# Trier selon la différence absolue
binary_results["abs_difference"] = (
    binary_results["difference"].abs()
)

binary_results = binary_results.sort_values(
    "abs_difference",
    ascending=False
)

# Sauvegarder le tableau
binary_results.to_csv(
    output_dir / "binary_target_rates.csv",
    index=False
)

print("\nVariables binaires les plus différenciantes :")
print(
    binary_results[
        [
            "variable",
            "target_rate_0",
            "target_rate_1",
            "difference"
        ]
    ].head(10).to_string(index=False)
)

# ------------------------------------------------------------
# Graphique
# ------------------------------------------------------------

top_binary = binary_results.head(10).copy()

plt.figure(figsize=(11, 7))

plt.barh(
    top_binary["variable"],
    top_binary["difference"]
)

plt.axvline(0, linewidth=1)

plt.title(
    "Binary Variables with the Largest Difference in Target Rate"
)

plt.xlabel(
    "Difference in target=1 rate (percentage points)"
)

plt.ylabel("Variable")

plt.grid(axis="x", alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir / "02_binary_target_effect.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# 5. VARIABLES CATEGORIELLES : TAUX DE TARGET = 1
# ------------------------------------------------------------

categorical_cols = [
    col for col in df.columns
    if col.endswith("_cat")
]

print("\n" + "=" * 60)
print("ANALYSE DES VARIABLES CATEGORIELLES")
print("=" * 60)

categorical_results = []

for col in categorical_cols:

    grouped = (
        df.groupby(col)[target]
        .agg(["mean", "count"])
        .reset_index()
    )

    grouped["target_rate"] = grouped["mean"] * 100

    # différence entre le taux maximum et minimum
    difference = (
        grouped["target_rate"].max()
        - grouped["target_rate"].min()
    )

    categorical_results.append({
        "variable": col,
        "n_categories": grouped[col].nunique(),
        "min_target_rate": grouped["target_rate"].min(),
        "max_target_rate": grouped["target_rate"].max(),
        "difference": difference
    })

categorical_results = pd.DataFrame(
    categorical_results
)

categorical_results = categorical_results.sort_values(
    "difference",
    ascending=False
)

categorical_results.to_csv(
    output_dir / "categorical_target_rates.csv",
    index=False
)

print("\nVariables catégorielles les plus différenciantes :")
print(
    categorical_results.head(10).to_string(index=False)
)


# ------------------------------------------------------------
# 6. GRAPHIQUE DES VARIABLES CATEGORIELLES
# ------------------------------------------------------------

top_cat = categorical_results.head(8)

plt.figure(figsize=(11, 7))

plt.barh(
    top_cat["variable"],
    top_cat["difference"]
)

plt.title(
    "Categorical Variables with the Largest Target-Rate Difference"
)

plt.xlabel(
    "Difference between maximum and minimum target rate (percentage points)"
)

plt.ylabel("Variable")

plt.grid(axis="x", alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir / "03_categorical_target_effect.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# 7. VARIABLES NUMÉRIQUES : CORRÉLATION DE SPEARMAN
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ANALYSE DES VARIABLES NUMÉRIQUES")
print("=" * 60)

# Toutes les colonnes numériques
numeric_cols = df.select_dtypes(
    include=np.number
).columns.tolist()

# Retirer target
numeric_features = [
    col for col in numeric_cols
    if col != target
]

# Corrélation de Spearman
spearman_corr = (
    df[numeric_features + [target]]
    .corr(method="spearman")[target]
    .drop(target)
    .sort_values(key=abs, ascending=False)
)

print("\nVariables numériques avec la plus forte association monotone avec target :")
print(spearman_corr.head(10))


# ------------------------------------------------------------
# 8. TOP VARIABLES NUMÉRIQUES : TARGET RATE PAR QUANTILES
# ------------------------------------------------------------

top_numeric = spearman_corr.abs().head(6).index.tolist()

fig, axes = plt.subplots(
    2,
    3,
    figsize=(16, 9)
)

axes = axes.flatten()

for ax, col in zip(axes, top_numeric):

    temp = df[[col, target]].dropna().copy()

    # Création de 10 groupes de taille approximativement égale
    try:

        temp["bin"] = pd.qcut(
            temp[col],
            q=10,
            duplicates="drop"
        )

        grouped = (
            temp.groupby("bin", observed=True)[target]
            .mean()
            * 100
        )

        x = range(len(grouped))

        ax.plot(
            x,
            grouped.values,
            marker="o"
        )

        ax.set_title(col)
        ax.set_xlabel("Value quantile")
        ax.set_ylabel("Target = 1 rate (%)")
        ax.grid(alpha=0.3)

        ax.set_xticks(list(x))

        ax.set_xticklabels(
            [str(i + 1) for i in x]
        )

    except Exception as e:

        ax.text(
            0.5,
            0.5,
            f"Unable to plot\n{e}",
            ha="center",
            va="center"
        )

# Si moins de 6 variables
for ax in axes[len(top_numeric):]:
    ax.axis("off")

fig.suptitle(
    "Target Rate across Quantiles of the Most Relevant Numerical Variables",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    output_dir / "04_numeric_target_relationships.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# 9. MISSING VALUES : COMPARAISON AVEC TARGET
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSING VALUES ET TARGET")
print("=" * 60)

missing_results = []

for col in df.columns:

    if col == target:
        continue

    missing_mask = df[col].isna()

    if missing_mask.sum() > 0:

        rate_missing = (
            df.loc[missing_mask, target].mean() * 100
        )

        rate_present = (
            df.loc[~missing_mask, target].mean() * 100
        )

        difference = rate_missing - rate_present

        missing_results.append({
            "variable": col,
            "missing_percent": missing_mask.mean() * 100,
            "target_rate_missing": rate_missing,
            "target_rate_present": rate_present,
            "difference": difference
        })

missing_results = pd.DataFrame(missing_results)

if not missing_results.empty:

    missing_results["abs_difference"] = (
        missing_results["difference"].abs()
    )

    missing_results = missing_results.sort_values(
        "abs_difference",
        ascending=False
    )

    missing_results.to_csv(
        output_dir / "missing_target_relationship.csv",
        index=False
    )

    print(
        missing_results.head(10).to_string(index=False)
    )

    # Graphique
    top_missing = missing_results.head(10)

    plt.figure(figsize=(11, 7))

    plt.barh(
        top_missing["variable"],
        top_missing["difference"]
    )

    plt.axvline(0, linewidth=1)

    plt.title(
        "Variables with Missing Values: Difference in Target Rate"
    )

    plt.xlabel(
        "Target rate: missing - non-missing (percentage points)"
    )

    plt.ylabel("Variable")

    plt.grid(axis="x", alpha=0.3)

    plt.tight_layout()

    plt.savefig(
        output_dir / "05_missing_values_target_effect.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# 10. TAUX DE TARGET POUR LES VARIABLES BINAIRES
#     SOUS FORME DE TABLEAU COMPLET
# ------------------------------------------------------------

if not binary_results.empty:

    plt.figure(figsize=(12, 8))

    plot_data = binary_results.head(15).copy()

    y = np.arange(len(plot_data))

    plt.barh(
        y,
        plot_data["target_rate_1"]
    )

    plt.yticks(
        y,
        plot_data["variable"]
    )

    plt.xlabel("Target = 1 rate (%)")

    plt.ylabel("Variable")

    plt.title(
        "Target = 1 Rate for Binary Variables"
    )

    plt.grid(axis="x", alpha=0.3)

    plt.tight_layout()

    plt.savefig(
        output_dir / "06_binary_target_rates.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# 11. RÉSUMÉ FINAL
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("EDA TARGET TERMINÉE")
print("=" * 60)

print(f"\nLes graphiques ont été sauvegardés dans :")
print(output_dir)

print("\nFichiers créés :")

for file in sorted(output_dir.iterdir()):
    print(" -", file.name)

print("\nAnalyse terminée.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve
)


# ============================================================
# 1. PATHS
# ============================================================

input_file = r"C:\Users\User\Documents\SoSe26\Advanced Data Analytics\ADA Projekt\porto_seguro.csv"

output_dir = Path(
    r"C:\Users\User\Documents\SoSe26\Advanced Data Analytics\ADA Projekt\logistic_regression"
)

output_dir.mkdir(parents=True, exist_ok=True)


# ============================================================
# 2. LOAD DATASET
# ============================================================

print("=" * 70)
print("LOADING DATASET")
print("=" * 70)

df = pd.read_csv(input_file)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")


# ============================================================
# 3. X / y
# ============================================================

target = "target"

X = df.drop(columns=[target])
y = df[target]

print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget percentages:")
print(
    (y.value_counts(normalize=True) * 100).round(2)
)


# ============================================================
# 4. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n" + "=" * 70)
print("TRAIN / TEST SPLIT")
print("=" * 70)

print(f"Training observations : {len(X_train):,}")
print(f"Testing observations  : {len(X_test):,}")
```

```python
# ============================================================
# PORTO SEGURO - LOGISTIC REGRESSION BASELINE
# ============================================================



# ============================================================
# 5. IDENTIFY VARIABLE TYPES
# ============================================================

numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X_train.select_dtypes(
    exclude=np.number
).columns.tolist()

print("\n" + "=" * 70)
print("VARIABLE TYPES")
print("=" * 70)

print(f"Numerical variables   : {len(numeric_features)}")
print(f"Categorical variables : {len(categorical_features)}")


# ============================================================
# 6. CONVERT CATEGORICAL VARIABLES TO STRINGS
# ============================================================
#
# Some categorical Porto Seguro variables contain a mixture
# of integers and strings.
#
# We convert them BEFORE the pipeline so that OneHotEncoder
# receives only strings.
# ============================================================

if len(categorical_features) > 0:

    X_train[categorical_features] = (
        X_train[categorical_features].astype(str)
    )

    X_test[categorical_features] = (
        X_test[categorical_features].astype(str)
    )

print("\nCategorical variables converted to strings.")


# ============================================================
# 7. NUMERICAL PREPROCESSING
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


# ============================================================
# 8. CATEGORICAL PREPROCESSING
# ============================================================

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            )
        )
    ]
)


# ============================================================
# 9. COLUMN TRANSFORMER
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ============================================================
# 10. LOGISTIC REGRESSION
# ============================================================

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)


# ============================================================
# 11. COMPLETE PIPELINE
# ============================================================

model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            logistic_model
        )
    ]
)


# ============================================================
# 12. TRAIN MODEL
# ============================================================

print("\n" + "=" * 70)
print("TRAINING LOGISTIC REGRESSION")
print("=" * 70)

model.fit(
    X_train,
    y_train
)

print("Training completed successfully.")


# ============================================================
# 13. PREDICTIONS
# ============================================================

y_pred = model.predict(X_test)

y_proba = model.predict_proba(
    X_test
)[:, 1]


# ============================================================
# 14. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

report = classification_report(
    y_test,
    y_pred,
    digits=4
)

print(report)

with open(
    output_dir / "classification_report.txt",
    "w"
) as f:

    f.write(report)


# ============================================================
# 15. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)

plt.figure(figsize=(8, 6))

plt.imshow(cm)

plt.title(
    "Confusion Matrix - Logistic Regression"
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.xticks(
    [0, 1],
    ["0", "1"]
)

plt.yticks(
    [0, 1],
    ["0", "1"]
)

for i in range(2):

    for j in range(2):

        plt.text(
            j,
            i,
            f"{cm[i, j]:,}",
            ha="center",
            va="center"
        )

plt.colorbar()

plt.tight_layout()

plt.savefig(
    output_dir / "01_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# 16. ROC-AUC
# ============================================================

roc_auc = roc_auc_score(
    y_test,
    y_proba
)

print("\nROC-AUC:", round(roc_auc, 4))

fpr, tpr, _ = roc_curve(
    y_test,
    y_proba
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"Logistic Regression (AUC = {roc_auc:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "ROC Curve - Logistic Regression"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir / "02_roc_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# 17. PRECISION-RECALL CURVE
# ============================================================

pr_auc = average_precision_score(
    y_test,
    y_proba
)

print("PR-AUC:", round(pr_auc, 4))

precision_curve, recall_curve, _ = (
    precision_recall_curve(
        y_test,
        y_proba
    )
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall_curve,
    precision_curve,
    label=f"Logistic Regression (AP = {pr_auc:.4f})"
)

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title(
    "Precision-Recall Curve - Logistic Regression"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir / "03_precision_recall_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# 18. PREDICTED PROBABILITY DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 6))

plt.hist(
    y_proba[y_test.values == 0],
    bins=50,
    alpha=0.7,
    label="Actual target = 0"
)

plt.hist(
    y_proba[y_test.values == 1],
    bins=50,
    alpha=0.7,
    label="Actual target = 1"
)

plt.xlabel(
    "Predicted probability of target = 1"
)

plt.ylabel("Number of observations")

plt.title(
    "Distribution of Predicted Probabilities"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    output_dir / "04_predicted_probability_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# 19. MODEL COEFFICIENTS
# ============================================================

print("\n" + "=" * 70)
print("EXTRACTING MODEL COEFFICIENTS")
print("=" * 70)

feature_names = (
    model
    .named_steps["preprocessor"]
    .get_feature_names_out()
)

coefficients = (
    model
    .named_steps["classifier"]
    .coef_[0]
)

coefficients_df = pd.DataFrame({
    "feature": feature_names,
    "coefficient": coefficients
})

coefficients_df["odds_ratio"] = np.exp(
    coefficients_df["coefficient"]
)

coefficients_df = coefficients_df.sort_values(
    "coefficient"
)

coefficients_df.to_csv(
    output_dir / "logistic_regression_coefficients.csv",
    index=False
)

print("\nCoefficient table saved.")


# ============================================================
# 20. TOP POSITIVE COEFFICIENTS
# ============================================================

top_positive = (
    coefficients_df
    .sort_values(
        "coefficient",
        ascending=False
    )
    .head(15)
    .sort_values("coefficient")
)

plt.figure(figsize=(10, 8))

plt.barh(
    top_positive["feature"],
    top_positive["coefficient"]
)

plt.xlabel("Logistic Regression Coefficient")
plt.ylabel("Feature")

plt.title(
    "Top Positive Logistic Regression Coefficients"
)

plt.grid(
    axis="x",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    output_dir / "05_top_positive_coefficients.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# 21. TOP NEGATIVE COEFFICIENTS
# ============================================================

top_negative = (
    coefficients_df
    .sort_values(
        "coefficient",
        ascending=True
    )
    .head(15)
)

plt.figure(figsize=(10, 8))

plt.barh(
    top_negative["feature"],
    top_negative["coefficient"]
)

plt.xlabel("Logistic Regression Coefficient")
plt.ylabel("Feature")

plt.title(
    "Top Negative Logistic Regression Coefficients"
)

plt.grid(
    axis="x",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    output_dir / "06_top_negative_coefficients.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# 22. FINAL METRICS
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

results = pd.DataFrame({
    "Model": [
        "Logistic Regression"
    ],
    "Accuracy": [
        accuracy
    ],
    "Precision": [
        precision
    ],
    "Recall": [
        recall
    ],
    "F1": [
        f1
    ],
    "ROC_AUC": [
        roc_auc
    ],
    "PR_AUC": [
        pr_auc
    ]
})

print("\n" + "=" * 70)
print("FINAL MODEL RESULTS")
print("=" * 70)

print(
    results.to_string(index=False)
)

results.to_csv(
    output_dir / "logistic_regression_results.csv",
    index=False
)


# ============================================================
# 23. SAVE PREDICTIONS
# ============================================================

predictions = pd.DataFrame({
    "actual_target": y_test.values,
    "predicted_target": y_pred,
    "predicted_probability": y_proba
})

predictions.to_csv(
    output_dir / "logistic_regression_predictions.csv",
    index=False
)


# ============================================================
# 24. LIST OUTPUT FILES
# ============================================================

print("\n" + "=" * 70)
print("LOGISTIC REGRESSION COMPLETED")
print("=" * 70)

print("\nAll results were saved in:")
print(output_dir)

print("\nGenerated files:")

for file in sorted(output_dir.iterdir()):
    print(" -", file.name)
```

```python
# ============================================================
# XGBOOST - RANDOMIZED SEARCH WITH CONTROLLED CV FOLDS
# ============================================================
#
# Workflow:
#
# Raw training data
#       ↓
# Preprocessing
#       ↓
# Numeric feature matrix
#       ↓
# Fixed Stratified CV folds
#       ↓
# RandomizedSearchCV
#       ↓
# Top 5 configurations
#
# IMPORTANT:
# The test set is NOT used anywhere in this experiment.
#
# The saved CV folds will later be reused for the
# scale_pos_weight experiment.
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import json
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from xgboost import XGBClassifier

from sklearn.model_selection import (
    RandomizedSearchCV,
    StratifiedKFold
)

from sklearn.metrics import (
    make_scorer,
    average_precision_score
)


# ============================================================
# 2. OUTPUT DIRECTORY
# ============================================================

output_dir = "XGBoost_RandomizedSearch_results"

os.makedirs(
    output_dir,
    exist_ok=True
)

print("=" * 70)
print("XGBOOST RANDOMIZED SEARCH")
print("=" * 70)


# ============================================================
# 3. CHECK INPUT DATA
# ============================================================

print("\nChecking input data...")

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)


# ============================================================
# 4. USE THE EXISTING PREPROCESSOR
# ============================================================
#
# We previously created a ColumnTransformer for the project.
#
# The important point is that XGBoost must NOT receive
# raw string columns.
#
# The preprocessor must convert categorical variables
# into numerical features.
#
# We first identify the existing preprocessor.
# ============================================================


if "preprocessor" not in globals():

    raise NameError(
        "The variable 'preprocessor' does not exist. "
        "Please run the preprocessing cell first."
    )


print("\nPreprocessor found:")
print(type(preprocessor))


# ============================================================
# 5. FIT PREPROCESSOR ONLY ON TRAINING DATA
# ============================================================
#
# This avoids using the test set.
#
# The preprocessing is learned exclusively from X_train.
# ============================================================

print("\nFitting preprocessing on X_train...")

X_train_processed = preprocessor.fit_transform(
    X_train
)


# ============================================================
# 6. CONVERT TO NUMERIC DATAFRAME
# ============================================================
#
# XGBoost requires numerical values.
#
# Depending on the preprocessing configuration,
# the result can be a NumPy array or sparse matrix.
# We convert it into a numerical matrix.
# ============================================================

if hasattr(
    X_train_processed,
    "toarray"
):

    X_train_processed = X_train_processed.toarray()


X_train_processed = np.asarray(
    X_train_processed,
    dtype=np.float32
)


print(
    "Processed training shape:",
    X_train_processed.shape
)

print(
    "Processed data type:",
    X_train_processed.dtype
)


# ============================================================
# 7. FINAL DATA TYPE CHECK
# ============================================================

if not np.issubdtype(
    X_train_processed.dtype,
    np.number
):

    raise TypeError(
        "The processed training data is still not numerical."
    )


print(
    "\nSUCCESS: All processed features are numerical."
)


# ============================================================
# 8. CREATE FIXED STRATIFIED CV FOLDS
# ============================================================
#
# We create the folds ONCE.
#
# These exact folds will be saved and reused later for:
#
#   1. Unweighted XGBoost
#   2. Weighted XGBoost
#
# This gives us a controlled comparison.
# ============================================================

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)


cv_splits = list(
    cv.split(
        X_train_processed,
        y_train
    )
)


print("\nCV folds created:")
print("Number of folds:", len(cv_splits))


# ============================================================
# 9. SAVE THE CV FOLDS
# ============================================================

folds_path = os.path.join(
    output_dir,
    "cv_folds.pkl"
)

with open(
    folds_path,
    "wb"
) as f:

    pickle.dump(
        cv_splits,
        f
    )


print(
    "\nCV folds saved to:",
    folds_path
)


# ============================================================
# 10. DEFINE XGBOOST BASE MODEL
# ============================================================

xgb_model = XGBClassifier(

    objective="binary:logistic",

    eval_metric="logloss",

    random_state=42,

    n_jobs=1
)


# ============================================================
# 11. DEFINE HYPERPARAMETER SEARCH SPACE
# ============================================================
#
# These parameters control:
#
# n_estimators       = number of trees
# max_depth          = tree depth
# learning_rate      = contribution of each tree
# subsample          = proportion of observations used
# colsample_bytree   = proportion of features used
# min_child_weight   = minimum child weight
# gamma              = minimum loss reduction
# reg_alpha          = L1 regularization
# reg_lambda         = L2 regularization
# ============================================================

param_distributions = {

    "n_estimators": [
        100,
        200,
        300,
        400
    ],

    "max_depth": [
        3,
        4,
        5,
        6
    ],

    "learning_rate": [
        0.01,
        0.05,
        0.1,
        0.2
    ],

    "subsample": [
        0.7,
        0.8,
        0.9,
        1.0
    ],

    "colsample_bytree": [
        0.7,
        0.8,
        0.9,
        1.0
    ],

    "min_child_weight": [
        1,
        3,
        5,
        10
    ],

    "gamma": [
        0,
        0.1,
        0.3,
        0.5
    ],

    "reg_alpha": [
        0,
        0.01,
        0.1
    ],

    "reg_lambda": [
        1,
        2,
        5,
        10
    ]
}


# ============================================================
# 12. DEFINE AVERAGE PRECISION SCORER
# ============================================================

ap_scorer = make_scorer(
    average_precision_score,
    response_method="predict_proba"
)


# ============================================================
# 13. RANDOMIZED SEARCH
# ============================================================
#
# n_iter controls how many combinations are tested.
#
# 36 combinations × 3 folds = 108 model fits.
#
# This is intentionally moderate so that the experiment
# remains computationally reasonable.
# ============================================================

random_search = RandomizedSearchCV(

    estimator=xgb_model,

    param_distributions=param_distributions,

    n_iter=36,

    scoring=ap_scorer,

    cv=cv_splits,

    refit=False,

    random_state=42,

    n_jobs=-1,

    verbose=2,

    return_train_score=True
)


# ============================================================
# 14. RUN RANDOMIZED SEARCH
# ============================================================

print("\n")
print("=" * 70)
print("STARTING RANDOMIZED SEARCH")
print("=" * 70)

random_search.fit(
    X_train_processed,
    y_train
)


print("\n")
print("=" * 70)
print("RANDOMIZED SEARCH COMPLETED")
print("=" * 70)


# ============================================================
# 15. SAVE COMPLETE CV RESULTS
# ============================================================

results_df = pd.DataFrame(
    random_search.cv_results_
)


results_df = results_df.sort_values(
    by="mean_test_score",
    ascending=False
)


results_csv = os.path.join(
    output_dir,
    "randomized_search_all_results.csv"
)


results_df.to_csv(
    results_csv,
    index=False
)


print(
    "\nComplete results saved to:",
    results_csv
)


# ============================================================
# 16. EXTRACT TOP 5 MODELS
# ============================================================

top5 = results_df.head(5).copy()


top5 = top5.reset_index(
    drop=True
)


top5.insert(
    0,
    "Rank",
    range(
        1,
        len(top5) + 1
    )
)


# ============================================================
# 17. DISPLAY TOP 5
# ============================================================

print("\n")
print("=" * 70)
print("TOP 5 XGBOOST CONFIGURATIONS")
print("=" * 70)


for _, row in top5.iterrows():

    print(
        f"\nMODEL RANK {int(row['Rank'])}"
    )

    print(
        f"CV Average Precision : "
        f"{row['mean_test_score']:.6f}"
    )

    print(
        f"CV AP Std            : "
        f"{row['std_test_score']:.6f}"
    )

    print("\nHyperparameters:")

    params = row["params"]

    for key, value in params.items():

        print(
            f"  {key}: {value}"
        )


# ============================================================
# 18. SAVE TOP 5 PARAMETERS
# ============================================================

top5_models = []


for _, row in top5.iterrows():

    top5_models.append({

        "rank":
            int(row["Rank"]),

        "cv_average_precision":
            float(row["mean_test_score"]),

        "cv_ap_std":
            float(row["std_test_score"]),

        "hyperparameters":
            row["params"]
    })


top5_json = os.path.join(
    output_dir,
    "top5_models_parameters.json"
)


with open(
    top5_json,
    "w"
) as f:

    json.dump(
        top5_models,
        f,
        indent=4
    )


print(
    "\nTop 5 configurations saved to:",
    top5_json
)


# ============================================================
# 19. CREATE A CLEAN PORTFOLIO TABLE
# ============================================================
#
# We keep the most informative hyperparameters instead of
# putting every parameter into the table.
#
# The complete hyperparameter information remains available
# in top5_models_parameters.json.
# ============================================================

portfolio_df = pd.DataFrame({

    "Rank":
        top5["Rank"],

    "AP":
        top5["mean_test_score"].round(5),

    "AP Std":
        top5["std_test_score"].round(5),

    "Trees":
        top5["param_n_estimators"],

    "Depth":
        top5["param_max_depth"],

    "LR":
        top5["param_learning_rate"],

    "Subsample":
        top5["param_subsample"],

    "Colsample":
        top5["param_colsample_bytree"],

    "MinChild":
        top5["param_min_child_weight"],

    "Gamma":
        top5["param_gamma"]
})


print("\n")
print("=" * 70)
print("PORTFOLIO TABLE")
print("=" * 70)

print(
    portfolio_df.to_string(
        index=False
    )
)


# ============================================================
# 20. SAVE PORTFOLIO TABLE AS CSV
# ============================================================

portfolio_csv = os.path.join(
    output_dir,
    "top5_models_portfolio_table.csv"
)


portfolio_df.to_csv(
    portfolio_csv,
    index=False
)


# ============================================================
# 21. CREATE TABLE IMAGE
# ============================================================

fig, ax = plt.subplots(
    figsize=(15, 4.5)
)

ax.axis("off")


table = ax.table(

    cellText=portfolio_df.values,

    colLabels=portfolio_df.columns,

    cellLoc="center",

    loc="center"
)


table.auto_set_font_size(
    False
)

table.set_fontsize(
    10
)

table.scale(
    1,
    2.2
)


plt.title(
    "Top 5 XGBoost Configurations – Randomized Search",
    fontsize=14,
    pad=20
)


plt.tight_layout()


table_png = os.path.join(
    output_dir,
    "top5_models_portfolio_table.png"
)


plt.savefig(
    table_png,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 22. PERFORMANCE PLOT
# ============================================================

fig, ax = plt.subplots(
    figsize=(9, 5)
)


ranks = top5["Rank"]

scores = top5[
    "mean_test_score"
]


ax.bar(
    ranks,
    scores
)


ax.set_xlabel(
    "Model Rank"
)

ax.set_ylabel(
    "Cross-Validation Average Precision"
)

ax.set_title(
    "Top 5 XGBoost Models - Cross-Validation AP"
)


ax.set_xticks(
    ranks
)


plt.tight_layout()


performance_png = os.path.join(
    output_dir,
    "top5_models_AP.png"
)


plt.savefig(
    performance_png,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 23. SAVE EXPERIMENT INFORMATION
# ============================================================

experiment_info = {

    "model":
        "XGBoost",

    "search_method":
        "RandomizedSearchCV",

    "n_iter":
        36,

    "cv_folds":
        3,

    "random_state":
        42,

    "metric":
        "Average Precision",

    "test_set_used":
        False,

    "preprocessing_applied":
        True,

    "cv_folds_saved":
        True
}


experiment_json = os.path.join(
    output_dir,
    "experiment_configuration.json"
)


with open(
    experiment_json,
    "w"
) as f:

    json.dump(
        experiment_info,
        f,
        indent=4
    )


# ============================================================
# 24. FINAL CHECK
# ============================================================

print("\n")
print("=" * 70)
print("EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nSaved files:")

print(
    "1. cv_folds.pkl"
)

print(
    "2. randomized_search_all_results.csv"
)

print(
    "3. top5_models_parameters.json"
)

print(
    "4. top5_models_portfolio_table.csv"
)

print(
    "5. top5_models_portfolio_table.png"
)

print(
    "6. top5_models_AP.png"
)

print(
    "7. experiment_configuration.json"
)

print("\nTest set used: NO")

print(
    "\nThe saved CV folds can now be reused "
    "for the controlled scale_pos_weight experiment."
)
```

```python
# ============================================================
# WEIGHTED XGBOOST - EVALUATION OF THE TOP 5 MODELS
# Using the SAME CV folds as the non-weighted RandomizedSearchCV
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.metrics import average_precision_score, roc_auc_score
from xgboost import XGBClassifier


# ============================================================
# 1. Retrieve the Top 5 Models from RandomizedSearchCV
# ============================================================

results = pd.DataFrame(random_search.cv_results_)

top5 = (
    results
    .sort_values("rank_test_score")
    .head(5)
    .copy()
    .reset_index(drop=True)
)

print("Top 5 models selected from RandomizedSearchCV:\n")

for i, row in top5.iterrows():
    print(f"\n{'='*60}")
    print(f"MODEL RANK {i+1}")
    print(f"{'='*60}")

    print(f"Non-weighted CV Average Precision : "
          f"{row['mean_test_score']:.6f}")

    print("Hyperparameters:")
    for key, value in row["params"].items():
        print(f"  {key}: {value}")


# ============================================================
# 2. Evaluate the 5 Models with Class Weighting
# ============================================================

weighted_results = []

# Store fold-level results for additional analysis
fold_results = []


for model_rank, row in enumerate(top5.itertuples(index=False), start=1):

    params = row.params

    print(f"\n{'='*70}")
    print(f"WEIGHTED MODEL RANK {model_rank}")
    print(f"{'='*70}")

    model_fold_ap = []
    model_fold_auc = []

    for fold_number, (train_idx, val_idx) in enumerate(cv_splits, start=1):

        # ----------------------------------------------------
        # Split the SAME folds used by RandomizedSearchCV
        # ----------------------------------------------------

        X_fold_train = X_train.iloc[train_idx]
        X_fold_val = X_train.iloc[val_idx]

        y_fold_train = y_train.iloc[train_idx]
        y_fold_val = y_train.iloc[val_idx]


        # ----------------------------------------------------
        # Calculate scale_pos_weight ONLY from the
        # training portion of the current fold
        # ----------------------------------------------------

        n_negative = (y_fold_train == 0).sum()
        n_positive = (y_fold_train == 1).sum()

        scale_pos_weight = n_negative / n_positive


        # ----------------------------------------------------
        # Clone the preprocessor
        # This prevents information leakage between folds
        # ----------------------------------------------------

        fold_preprocessor = clone(preprocessor)

        X_fold_train_transformed = fold_preprocessor.fit_transform(
            X_fold_train,
            y_fold_train
        )

        X_fold_val_transformed = fold_preprocessor.transform(
            X_fold_val
        )


        # ----------------------------------------------------
        # Create the weighted XGBoost model
        # ----------------------------------------------------

        weighted_model = XGBClassifier(
            **params,
            scale_pos_weight=scale_pos_weight,
            random_state=42,
            eval_metric="logloss"
        )


        # ----------------------------------------------------
        # Train on the weighted training fold
        # ----------------------------------------------------

        weighted_model.fit(
            X_fold_train_transformed,
            y_fold_train
        )


        # ----------------------------------------------------
        # Predict probabilities on the UNWEIGHTED
        # validation fold
        # ----------------------------------------------------

        y_val_proba = weighted_model.predict_proba(
            X_fold_val_transformed
        )[:, 1]


        # ----------------------------------------------------
        # Calculate validation metrics
        # ----------------------------------------------------

        fold_ap = average_precision_score(
            y_fold_val,
            y_val_proba
        )

        fold_auc = roc_auc_score(
            y_fold_val,
            y_val_proba
        )


        model_fold_ap.append(fold_ap)
        model_fold_auc.append(fold_auc)


        fold_results.append({
            "Model": model_rank,
            "Fold": fold_number,
            "Scale_Pos_Weight": scale_pos_weight,
            "Average_Precision": fold_ap,
            "ROC_AUC": fold_auc
        })


        print(
            f"Fold {fold_number}: "
            f"AP = {fold_ap:.6f} | "
            f"ROC-AUC = {fold_auc:.6f} | "
            f"scale_pos_weight = {scale_pos_weight:.3f}"
        )


    # --------------------------------------------------------
    # Calculate mean and standard deviation across folds
    # --------------------------------------------------------

    mean_ap = np.mean(model_fold_ap)
    std_ap = np.std(model_fold_ap)

    mean_auc = np.mean(model_fold_auc)
    std_auc = np.std(model_fold_auc)


    weighted_results.append({
        "Rank": model_rank,
        "NonWeighted_CV_AP": row.mean_test_score,
        "Weighted_CV_AP": mean_ap,
        "Weighted_CV_AP_Std": std_ap,
        "Weighted_CV_ROC_AUC": mean_auc,
        "Weighted_CV_ROC_AUC_Std": std_auc
    })


# ============================================================
# 3. Create Results Table
# ============================================================

weighted_results_df = pd.DataFrame(weighted_results)

print("\n\n")
print("=" * 90)
print("TOP 5 MODELS - WEIGHTED VS NON-WEIGHTED")
print("=" * 90)

print(weighted_results_df.round(6))


# ============================================================
# 4. Save Numerical Results
# ============================================================

weighted_results_df.to_csv(
    "xgboost_top5_weighted_results.csv",
    index=False
)

fold_results_df = pd.DataFrame(fold_results)

fold_results_df.to_csv(
    "xgboost_weighted_fold_results.csv",
    index=False
)

print("\nResults saved:")
print("- xgboost_top5_weighted_results.csv")
print("- xgboost_weighted_fold_results.csv")


# ============================================================
# 5. Plot Weighted CV Average Precision
#    Mean AP ± Standard Deviation
# ============================================================

plt.figure(figsize=(12, 7))

x = weighted_results_df["Rank"]
y = weighted_results_df["Weighted_CV_AP"]
yerr = weighted_results_df["Weighted_CV_AP_Std"]

plt.errorbar(
    x,
    y,
    yerr=yerr,
    marker="o",
    markersize=8,
    linewidth=2,
    capsize=6
)

plt.xlabel("Model Rank")
plt.ylabel("Weighted Cross-Validation Average Precision")
plt.title(
    "Top 5 XGBoost Models - Weighted CV Average Precision"
)

plt.xticks(x)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    "top5_weighted_average_precision.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 6. Plot Weighted CV ROC-AUC
# ============================================================

plt.figure(figsize=(12, 7))

x = weighted_results_df["Rank"]
y = weighted_results_df["Weighted_CV_ROC_AUC"]
yerr = weighted_results_df["Weighted_CV_ROC_AUC_Std"]

plt.errorbar(
    x,
    y,
    yerr=yerr,
    marker="o",
    markersize=8,
    linewidth=2,
    capsize=6
)

plt.xlabel("Model Rank")
plt.ylabel("Weighted Cross-Validation ROC-AUC")
plt.title(
    "Top 5 XGBoost Models - Weighted CV ROC-AUC"
)

plt.xticks(x)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    "top5_weighted_roc_auc.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 7. Compare Non-Weighted vs Weighted Average Precision
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(
    weighted_results_df["Rank"],
    weighted_results_df["NonWeighted_CV_AP"],
    marker="o",
    linewidth=2,
    label="Non-weighted"
)

plt.plot(
    weighted_results_df["Rank"],
    weighted_results_df["Weighted_CV_AP"],
    marker="o",
    linewidth=2,
    label="Weighted"
)

plt.xlabel("Model Rank")
plt.ylabel("Cross-Validation Average Precision")

plt.title(
    "Top 5 XGBoost Models - Non-Weighted vs Weighted CV AP"
)

plt.xticks(weighted_results_df["Rank"])
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    "top5_nonweighted_vs_weighted_AP.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 8. Calculate the Change in Average Precision
# ============================================================

weighted_results_df["AP_Change"] = (
    weighted_results_df["Weighted_CV_AP"]
    - weighted_results_df["NonWeighted_CV_AP"]
)

weighted_results_df["AP_Change_Percent"] = (
    weighted_results_df["AP_Change"]
    / weighted_results_df["NonWeighted_CV_AP"]
) * 100


print("\n")
print("=" * 90)
print("CHANGE IN AVERAGE PRECISION AFTER CLASS WEIGHTING")
print("=" * 90)

print(
    weighted_results_df[
        [
            "Rank",
            "NonWeighted_CV_AP",
            "Weighted_CV_AP",
            "AP_Change",
            "AP_Change_Percent"
        ]
    ].round(6)
)


# ============================================================
# 9. Save Final Comparison Table
# ============================================================

weighted_results_df.to_csv(
    "xgboost_top5_weighted_comparison.csv",
    index=False
)

print(
    "\nFinal comparison saved as: "
    "xgboost_top5_weighted_comparison.csv"
)
```

```python
# ============================================================
# XGBOOST - TARGETED GRID SEARCH
# ============================================================
#
# This GridSearchCV performs an exhaustive search over a
# restricted hyperparameter space identified from the previous
# RandomizedSearchCV.
#
# IMPORTANT:
# - The test set is NOT used.
# - The same X_train / y_train are used.
# - The same controlled CV folds are reused.
# - Average Precision is used as the primary metric because
#   the Porto Seguro target is highly imbalanced.
# ============================================================

import os
import json
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import average_precision_score, roc_auc_score


# ============================================================
# 1. OUTPUT DIRECTORY
# ============================================================

output_dir = "XGBoost_GridSearch_results"
os.makedirs(output_dir, exist_ok=True)

print("=" * 70)
print("TARGETED XGBOOST GRID SEARCH")
print("=" * 70)


# ============================================================
# 2. LOAD THE CONTROLLED CV FOLDS
# ============================================================
#
# These are the SAME folds used during RandomizedSearchCV.
# This makes the comparison between RandomizedSearchCV and
# GridSearchCV more consistent.
# ============================================================

folds_path = os.path.join(
    "XGBoost_RandomizedSearch_results",
    "cv_folds.pkl"
)

with open(folds_path, "rb") as f:
    cv_splits = pickle.load(f)

print("\nControlled CV folds loaded.")
print("Number of folds:", len(cv_splits))


# ============================================================
# 3. CHECK REQUIRED VARIABLES
# ============================================================

required_variables = [
    "X_train",
    "y_train",
    "preprocessor"
]

missing_variables = [
    variable
    for variable in required_variables
    if variable not in globals()
]

if missing_variables:

    raise NameError(
        "The following variables are missing: "
        + ", ".join(missing_variables)
        + "\nPlease recreate the training data and preprocessor "
          "before running this cell."
    )


# ============================================================
# 4. PREPROCESS THE TRAINING DATA
# ============================================================
#
# XGBoost cannot directly process raw string categorical
# variables.
#
# The preprocessor converts categorical variables into
# numerical features.
#
# IMPORTANT:
# The test set is NOT touched here.
# ============================================================

print("\nFitting preprocessor on X_train...")

X_train_processed = preprocessor.fit_transform(X_train)

print("Training data successfully preprocessed.")
print("Processed X_train shape:", X_train_processed.shape)


# ============================================================
# 5. DEFINE THE BASE XGBOOST MODEL
# ============================================================

xgb_model = XGBClassifier(
    objective="binary:logistic",
    eval_metric="logloss",
    tree_method="hist",
    random_state=42,
    n_jobs=-1
)


# ============================================================
# 6. TARGETED GRID
# ============================================================
#
# The values below are deliberately restricted to promising
# regions identified by the previous RandomizedSearchCV.
#
# This avoids an unnecessarily large exhaustive search.
# ============================================================

param_grid = {

    "n_estimators": [
        200,
        300,
        400
    ],

    "max_depth": [
        3,
        4
    ],

    "learning_rate": [
        0.05,
        0.10
    ],

    "subsample": [
        0.7,
        0.9,
        1.0
    ],

    "colsample_bytree": [
        0.8,
        0.9,
        1.0
    ],

    "min_child_weight": [
        3,
        5
    ],

    "gamma": [
        0.0,
        0.1,
        0.3
    ],

    "reg_alpha": [
        0.0,
        0.01,
        0.1
    ],

    "reg_lambda": [
        1,
        2,
        5
    ]
}


# ============================================================
# 7. CALCULATE NUMBER OF COMBINATIONS
# ============================================================

number_of_combinations = np.prod(
    [
        len(values)
        for values in param_grid.values()
    ]
)

number_of_fits = (
    number_of_combinations
    * len(cv_splits)
)

print("\nGrid Search configuration:")
print("----------------------------------------")

for parameter, values in param_grid.items():
    print(f"{parameter}: {values}")

print("----------------------------------------")
print(
    "Total parameter combinations:",
    number_of_combinations
)

print(
    "Total model fits:",
    number_of_fits
)

print(
    "\nWARNING: This is an exhaustive search."
)

print(
    "Depending on the hardware, execution may take "
    "considerable time."
)


# ============================================================
# 8. GRID SEARCH
# ============================================================
#
# Average Precision is used because the positive class is rare.
# The same controlled CV folds are reused.
# ============================================================

grid_search = GridSearchCV(

    estimator=xgb_model,

    param_grid=param_grid,

    scoring="average_precision",

    cv=cv_splits,

    n_jobs=-1,

    verbose=2,

    return_train_score=True
)


print("\n")
print("=" * 70)
print("STARTING GRID SEARCH")
print("=" * 70)

grid_search.fit(
    X_train_processed,
    y_train
)

print("\nGrid Search completed.")


# ============================================================
# 9. BEST MODEL
# ============================================================

print("\n")
print("=" * 70)
print("BEST GRID SEARCH MODEL")
print("=" * 70)

print(
    "\nBest CV Average Precision:",
    grid_search.best_score_
)

print(
    "\nBest Hyperparameters:"
)

for parameter, value in grid_search.best_params_.items():
    print(f"  {parameter}: {value}")


# ============================================================
# 10. SAVE BEST PARAMETERS
# ============================================================

best_model_information = {

    "best_cv_average_precision":
        float(grid_search.best_score_),

    "best_hyperparameters":
        grid_search.best_params_
}

with open(
    os.path.join(
        output_dir,
        "best_gridsearch_model.json"
    ),
    "w"
) as f:

    json.dump(
        best_model_information,
        f,
        indent=4
    )


# ============================================================
# 11. EXTRACT COMPLETE RESULTS
# ============================================================

results = pd.DataFrame(
    grid_search.cv_results_
)

results = results.sort_values(
    by="rank_test_score"
)

print("\nTop 10 Grid Search models:")

display(
    results[
        [
            "rank_test_score",
            "mean_test_score",
            "std_test_score",
            "mean_train_score",
            "params"
        ]
    ].head(10)
)


# ============================================================
# 12. SAVE COMPLETE RESULTS
# ============================================================

results.to_csv(
    os.path.join(
        output_dir,
        "gridsearch_all_results.csv"
    ),
    index=False
)


# ============================================================
# 13. CREATE A CLEAN TOP-10 TABLE
# ============================================================
#
# Only the most informative hyperparameters are displayed.
# This keeps the table readable for the portfolio.
# ============================================================

top10 = results.head(10).copy()

top10_table = pd.DataFrame({

    "Rank":
        top10["rank_test_score"].values,

    "CV AP":
        top10["mean_test_score"].values,

    "AP Std":
        top10["std_test_score"].values,

    "Trees":
        top10["param_n_estimators"].values,

    "Depth":
        top10["param_max_depth"].values,

    "LR":
        top10["param_learning_rate"].values,

    "Subsample":
        top10["param_subsample"].values,

    "Colsample":
        top10["param_colsample_bytree"].values,

    "MinChild":
        top10["param_min_child_weight"].values,

    "Gamma":
        top10["param_gamma"].values,

    "Reg Alpha":
        top10["param_reg_alpha"].values,

    "Reg Lambda":
        top10["param_reg_lambda"].values
})


# ============================================================
# 14. SAVE TOP-10 TABLE AS CSV
# ============================================================

top10_table.to_csv(
    os.path.join(
        output_dir,
        "gridsearch_top10_models.csv"
    ),
    index=False
)


# ============================================================
# 15. CREATE PORTFOLIO-READY TABLE IMAGE
# ============================================================

fig, ax = plt.subplots(
    figsize=(18, 5)
)

ax.axis("off")

table = ax.table(

    cellText=top10_table.round(5).values,

    colLabels=top10_table.columns,

    cellLoc="center",

    loc="center"
)

table.auto_set_font_size(False)

table.set_fontsize(8)

table.scale(
    1,
    1.8
)

plt.title(
    "Top 10 XGBoost Configurations - Targeted Grid Search",
    fontsize=14,
    pad=20
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_dir,
        "gridsearch_top10_table.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 16. PERFORMANCE OF TOP 10 MODELS
# ============================================================

top10_plot = results.head(10).copy()

plt.figure(
    figsize=(10, 6)
)

plt.errorbar(

    range(
        1,
        len(top10_plot) + 1
    ),

    top10_plot["mean_test_score"],

    yerr=top10_plot["std_test_score"],

    marker="o",

    capsize=4
)

plt.xlabel(
    "Model Rank"
)

plt.ylabel(
    "Cross-Validation Average Precision"
)

plt.title(
    "Top 10 XGBoost Models - Grid Search"
)

plt.xticks(
    range(
        1,
        len(top10_plot) + 1
    )
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_dir,
        "gridsearch_top10_cv_ap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 17. SAVE GRID SEARCH OBJECT
# ============================================================

with open(
    os.path.join(
        output_dir,
        "grid_search_object.pkl"
    ),
    "wb"
) as f:

    pickle.dump(
        grid_search,
        f
    )


print("\n")
print("=" * 70)
print("GRID SEARCH FINISHED")
print("=" * 70)

print(
    "\nResults saved in:",
    output_dir
)

print(
    "\nFiles created:"
)

print(
    " - best_gridsearch_model.json"
)

print(
    " - gridsearch_all_results.csv"
)

print(
    " - gridsearch_top10_models.csv"
)

print(
    " - gridsearch_top10_table.png"
)

print(
    " - gridsearch_top10_cv_ap.png"
)

print(
    " - grid_search_object.pkl"
)
```
